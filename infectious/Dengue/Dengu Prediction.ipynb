```python
for i in range(data.shape[0]):
    if data.loc[i,'cases'] > 18000:
        data.loc[i,'labels'] = "Severe Risk"
    elif data.loc[i,'cases'] > 12000:
        data.loc[i, 'labels'] = "High Risk"
    elif data.loc[i,'cases'] > 6000:
        data.loc[i,'labels'] = "Moderate Risk"
    elif data.loc[i,'cases'] > 1000:
        data.loc[i,'labels'] = "Low Risk"
    else:
        data.loc[i,'labels'] = 'Minimal to No Risk'
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
data.labels = le.fit_transform(data.labels)
```

```python
X = data.iloc[:,:-1]
y = data.labels
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify= y, random_state=42)
```

```python
from sklearn.ensemble import RandomForestRegressor
reg = RandomForestRegressor()
```

```python
reg.fit(X_train,y_train)
```

```python
y_predict = reg.predict(X_test)
```

```python
reg.score(X_test,y_test)
```

```python
from sklearn.model_selection import cross_val_score
acc = cross_val_score(reg, X_test, y_test, cv=5).mean()
```

```python
print(acc)
```
