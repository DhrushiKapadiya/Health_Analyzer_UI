Importing the Dependencies

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn.metrics import accuracy_score
```

Data Collection and Analysis

PIMA Diabetes Dataset

```python
# loading the diabetes dataset to a pandas DataFrame
ks_dataset = pd.read_csv('kidney-stone-dataset.csv') 
ks_dataset
```

```python
# printing the first 5 rows of the dataset
ks_dataset.head()
```

```python
ks_dataset.info()
```

```python
# number of rows and Columns in this dataset
ks_dataset.shape
```

```python
# getting the statistical measures of the data
ks_dataset.describe()
```

```python
ks_dataset['target'].value_counts()
```

0 --> Non-KS

1 --> KS

```python
ks_dataset.groupby('target').mean()
```

```python
# separating the data and labels
X = ks_dataset.drop(columns = 'target', axis=1)
Y = ks_dataset['target']
```

```python
print(X)
```

```python
print(Y)
```

Train Test Split

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size = 0.15, stratify=Y, random_state=2)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

Training the Model

## Model : 

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
from sklearn.tree import DecisionTreeClassifier
classifier = DecisionTreeClassifier(criterion = 'entropy', random_state = 5)
classifier.fit(X_train, Y_train)
```

```python
y_pred = classifier.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score
cm = confusion_matrix(Y_test, y_pred)
print(cm)
accuracy_score(Y_test, y_pred)
```

```python
a = sc.fit_transform([[1.021,4.91,725,14.0,443,2.45]])

print(classifier.predict(a))
```

```python
input_data = (1.021,4.91,725,14.0,443,2.45)

# changing the input_data to numpy array
input_data_as_numpy_array = np.asarray(input_data)

# reshape the array as we are predicting for one instance
input_data_reshaped = input_data_as_numpy_array.reshape(1,-1)

prediction = classifier.predict(input_data_reshaped)
print(prediction)

if (prediction[0] == 0):
  print('The person has not Kidney Stone')
else:
  print('The person has Kidney Stone')
```

Saving the trained model

```python
import pickle
```

```python
filename = 'ks_model.sav'
pickle.dump(classifier, open(filename, 'wb'))
```
