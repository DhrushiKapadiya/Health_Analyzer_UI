Importing the Dependencies

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn.metrics import accuracy_score
```

Data Collection and Analysis

PIMA Diabetes Dataset

```python
# loading the diabetes dataset to a pandas DataFrame
stroke_dataset = pd.read_csv('stroke_data.csv') 
stroke_dataset
```

```python
stroke_dataset.dropna(inplace=True)
```

```python
stroke_dataset.info()
```

```python
# printing the first 5 rows of the dataset
stroke_dataset.head()
```

```python
# number of rows and Columns in this dataset
stroke_dataset.shape
```

```python
# getting the statistical measures of the data
stroke_dataset.describe()
```

```python
stroke_dataset['stroke'].value_counts()
```

0 --> Non-stroke

1 --> stroke

```python
stroke_dataset.groupby('stroke').mean()
```

```python
stroke_input = stroke_dataset.sample(20000)
```

```python
stroke_input
```

```python
stroke_input['stroke'].value_counts()
```

```python
# separating the data and labels
X = stroke_input.drop(columns = 'stroke', axis=1)
Y = stroke_input['stroke']
```

```python
print(X)
```

```python
print(Y)
```

Train Test Split

```python

X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size = 0.2, random_state=2)
```

```python
print(X.shape, X_train.shape, X_test.shape)
print(Y.shape, Y_train.shape, Y_test.shape)
```

Training the Model

```python
X_train
```

# Model :


```python
from sklearn.linear_model import LogisticRegression
regression = LogisticRegression(random_state = 5,max_iter=300,solver='liblinear')
regression.fit(X_train, Y_train)
```

```python
prediction = regression.predict(X_test)
prediction
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score
cm = confusion_matrix(Y_test, prediction)
print(cm)
accuracy_score(Y_test,prediction)
```

## Standard Scaler : 

```python
X_train
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
regression = LogisticRegression(random_state = 15,max_iter=300,solver='liblinear')
regression.fit(X_train, Y_train)
```

```python
from sklearn.tree import DecisionTreeClassifier
classifier = DecisionTreeClassifier(criterion = 'entropy', random_state = 5)
classifier.fit(X_train, Y_train)
```

```python
y_pred = classifier.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score
cm = confusion_matrix(Y_test, y_pred)
print(cm)
accuracy_score(Y_test, y_pred)
```

Making a Predictive System

```python
input_data = (1,63.0,0,1,1,4,1,228.69,36.3,1)

# changing the input_data to numpy array
input_data_as_numpy_array = np.asarray(input_data)

# reshape the array as we are predicting for one instance
input_data_reshaped = input_data_as_numpy_array.reshape(1,-1)

prediction = classifier.predict(input_data_reshaped)
print(prediction)

if (prediction[0] == 0):
  print('The person has not stroke')
else:
  print('The person has stroke')
```

Saving the trained model

```python
import pickle
```

```python
filename = 'stroke_model.sav'
pickle.dump(classifier, open(filename, 'wb'))
```
